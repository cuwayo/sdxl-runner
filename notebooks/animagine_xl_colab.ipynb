## 2. Launch the server

This loads the model lazily, starts a cloudflared quick tunnel, and prints the
public URL. Keep this cell running while you generate.

````python
from sdxl_runner.server.launch import launch_server

# `tunnel=True` starts a free cloudflared quick tunnel and prints a public URL.
# Use `share=True` for Gradio's own gradio.live link if the tunnel gives you trouble.
launch_server(port=8000, tunnel=True)
````

## 3. Generate from your machine

Copy the public URL from the banner above and run locally:

```bash
sdxl-runner generate \
  --url https://xxxx.trycloudflare.com \
  --quality --auto-size \
  -p "1girl, arima kana, oshi no ko, looking at viewer, smile, outdoors, night" \
  -o ./out
```
